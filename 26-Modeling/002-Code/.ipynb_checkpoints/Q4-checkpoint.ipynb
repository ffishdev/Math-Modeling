## 3 技术路线

问题三联合方案 → 多点架次并查集 → 不可拆分原子单元 → 穷尽枚举 2 组 / 3 组划分 → 固定时序下核算分组资源峰值 → 按缺口、冗余与均衡排序 → 推荐方案及图表分析。附加时刻平移只作为敏感性分析。

## 4 模型建立

**(1) 原子执行单元。** 令 $P$ 为问题三的运输架次集合，架次 $p$ 访问的服务区集合为 $A_p$。定义服务区之间的关系 $i\sim j \iff \exists p: i,j\in A_p$，其传递闭包把 15 个服务区划分为若干原子单元 $\mathcal U=\{U_1,\ldots,U_m\}$（并查集求连通分量）。题目要求“同一架次涉及的服务区同组”等价于：分区是原子单元的划分。

**(2) 分区。** $K$ 组划分 $\Pi=\{G_1,\ldots,G_K\}$，$G_k$ 为若干原子单元之并，$G_k\ne\emptyset$。$m$ 很小时可精确枚举（$K=2$ 有 $2^{m-1}-1$ 种，$K=3$ 有第二类 Stirling 数 $S(m,3)$ 种）。

**(3) 固定时序下的资源需求（区间图着色）。** 组 $G_k$ 的架次集合 $P_k=\{p:A_p\subseteq G_k\}$。机型 $g$ 的架次在时间轴上是区间 $[t_p^{st},t_p^{en})$；独立执行所需的机型 $g$ 无人机数

$$
n_{k,g}^{U}=\max_t |\{p\in P_k:g_p=g,\ t_p^{st}\le t<t_p^{en}\}|.
$$

即区间图的团数；区间图是完美图，按开始时刻贪心分配“最早空闲机”恰好用这么多架（分配结果就是分组甘特图的泳道）。电池同理，区间取 $[t_p^{st},t_p^{en}+t^{chg}(s_p))$（充满前不可再用）。

**(4) 中继资源。** 组 $G_k$ 在中继站 $s$ 的需求窗 = 该组架次在 $s$ 的需求区间之并；按问题三的会话规则（间隔小于“返回 + 周转 + 再去 + 建链”则留空悬停；超过单组件续航则无缝接替）重建会话，得到中继机占用区间 $[r^{dep},r^{end}+r^{turn})$ 与组件占用区间 $[r^{dep},r^{end}+t^{chg}(s))$，取并发峰值得 $n_k^R$、$n_k^C$。

**(5) 分区评价指标。**

- 配置规模：$N(\Pi)=\sum_k(\sum_g n_{k,g}^{U}+\sum_g n_{k,g}^{B}+n_k^R+n_k^C)$（并分资源类型报告）；
- 冗余：$\Delta(\Pi)=\sum_k n_k-n^{joint}$，其中 $n^{joint}$ 是全部架次联合执行（同一固定时序）所需的峰值资源——分区后各组峰值之和减去联合峰值，就是“资源不能跨组调配”造成的额外配置；
- 均衡：以架次作业时长为工作量，$\beta(\Pi)=\max_k|w_k/\sum_j w_j-1/K|$（越小越均衡），并报告质量、箱数、人口等份额；
- 缺口：$\gamma_r(\Pi)=\max\{0,\sum_k n_{k,r}-I_r\}$，$I_r$ 为库存。

推荐分区取字典序最优：先最小化加权缺口（无人机与中继机权重 3，电池与组件权重 1），再最小化冗余，再最小化均衡偏差；全部分区的指标表随结果导出，便于按其他偏好复选。

## 7 原子执行单元与图 1

用并查集把被同一运输架次串联的服务区合并，得到不可拆分的原子单元。图1在地图上画出“架次耦合网络”：节点为服务区，边为同一多点架次的连接（线宽 $\propto$ 共同架次数），同色凸包为一个原子单元。多点串联在问题二、三中显著减少了架次数与能耗，但代价是把大量服务区绑成一个巨大的执行单元—这是本问所 有结论的结构性前提。

```python
# 同架次服务区并查集
parent={a:a for a in AREA_IDS}; edges=Counter()
def find(x):
    if parent[x]!=x: parent[x]=find(parent[x])
    return parent[x]
def union(a,b):
    ra,rb=find(a),find(b)
    if ra!=rb: parent[rb]=ra
for ar in plan.areas:
    for a in ar[1:]:union(ar[0],a)
    for a,b in itertools.combinations(ar,2): edges[tuple(sorted((a,b)))]+=1
u_map=defaultdict(list)
for a in AREA_IDS:u_map[find(a)].append(a)
UNITS=sorted((tuple(v) for v in u_map.values()),key=lambda x:(-len(x),x[0]))
U_ID={a:i for i,u in enumerate(UNITS) for a in u}
unit_df=pd.DataFrame([{'原子单元':f'U{i+1}','服务区数':len(u),'服务区':'、'.join(u),
                       '运输架次':int(sum(set(ar).issubset(u) for ar in plan.areas)),
                       '货箱数':int(box_df['服务区'].isin(u).sum())} for i,u in enumerate(UNITS)])
assert sum(map(len,UNITS))==15 and all(len({U_ID[a] for a in ar})==1 for ar in plan.areas)
print(f'{sum(plan.停靠点.gt(1))} 个多点架次形成 {len(UNITS)} 个原子单元；K=2 候选 {2**(len(UNITS)-1)-1} 个')
display(unit_df)

# 图 1：原子单元与多点架次的耦合关系（DEM 仅作底图）
m=sio.loadmat(find_file('镇龙乡及周边30米DEM.mat'))
DEM=m['dem'].astype(float); DEM[DEM==float(np.ravel(m['nodata'])[0])]=np.nan
tr=m['transform'].ravel(); dx=float(tr[0]);dy=float(-tr[4]);lon0=float(tr[2]);lat0=float(tr[5])
extent=(lon0,lon0+DEM.shape[1]*dx,lat0-DEM.shape[0]*dy,lat0)
lon_min,lon_max=areas.lon.min()-.025,areas.lon.max()+.025
lat_min,lat_max=areas.lat.min()-.022,areas.lat.max()+.022
from scipy.spatial import ConvexHull

def base_map(ax):
    ax.imshow(DEM,extent=extent,origin='upper',cmap='gist_earth',alpha=.55,zorder=0)
    ax.set_xlim(lon_min,lon_max);ax.set_ylim(lat_min,lat_max)
    ax.set_xlabel('经度 (°E)');ax.set_ylabel('纬度 (°N)')
    ax.set_aspect(1/math.cos(math.radians((lat_min+lat_max)/2)))
    ax.scatter([LON_O],[LAT_O],marker='*',s=170,c=PAL['gold'],edgecolors=PAL['ink'],zorder=8)
    ax.annotate('O01',(LON_O,LAT_O),xytext=(5,4),textcoords='offset points',fontsize=8)
def hull_patch(ax,ids,color):
    xy=areas.loc[list(ids),['lon','lat']].to_numpy(float)
    if len(xy)>=3:
        h=ConvexHull(xy)
        ax.add_patch(Polygon(xy[h.vertices],closed=True,facecolor=color,edgecolor=color,alpha=.13,lw=1.3,zorder=2))
    elif len(xy)==2:
        ax.plot(xy[:,0],xy[:,1],color=color,lw=9,alpha=.18,zorder=2)
fig,ax=plt.subplots(figsize=(10,7));base_map(ax)
colors=[PAL['navy'],PAL['coral'],PAL['teal'],PAL['gold'],PAL['plum'],PAL['grey']]
for i,u in enumerate(UNITS):
    col=colors[i%len(colors)];hull_patch(ax,u,col)
    for a in u:
        r=areas.loc[a];ax.scatter(r.lon,r.lat,s=120,color=col,edgecolors='white',zorder=5)
        ax.annotate(a,(r.lon,r.lat),xytext=(4,5),textcoords='offset points',fontsize=7.5)
for (a,b),n in edges.items():
    p=areas.loc[[a,b],['lon','lat']].to_numpy(float)
    ax.plot(p[:,0],p[:,1],color=PAL['ink'],lw=1.1+n*.3,alpha=.65,zorder=3)
for i,u in enumerate(UNITS):
    xy=areas.loc[list(u),['lon','lat']].to_numpy(float).mean(axis=0)
    ax.text(xy[0],xy[1]-.003,f'U{i+1}',ha='center',weight='bold',fontsize=8,color=colors[i%len(colors)])
ax.set_title('图 1  多点架次耦合与不可拆分原子单元')
savefig(fig,'图01_原子单元与架次耦合')
```

## 8 分组资源核算函数（固定时序、区间图着色）

`group_eval(area_set)` 对一个任务组返回机型无人机和电池的并发峰值及峰值时刻、可核查的资源泳道、中继会话与组件需求、工作量指标。`JOINT` 记录所有架次联合执行时的峰值，作为分组冗余的基准。

```python
def max_overlap(intervals):
    intervals=list(intervals)
    ev=sorted([(float(a),1) for a,b in intervals if b>a+1e-9]+
              [(float(b),-1) for a,b in intervals if b>a+1e-9],key=lambda x:(x[0],x[1]))
    cur=best=0;tbest=0.
    for t,d in ev:
        cur+=d
        if cur>best:best,tbest=cur,t
    return best,tbest

def lanes(intervals):
    # 区间图最少着色：每次分配当前最早空闲的泳道。
    free=[];res={}
    for a,b,tag in sorted(intervals,key=lambda x:(x[0],x[1],str(x[2]))):
        j=next((j for j,x in enumerate(free) if x<=a+1e-9),None)
        if j is None:free.append(b);j=len(free)-1
        else:free[j]=b
        res[tag]=j
    return res

def t_chg(soc,full):
    s=max(0.,min(1.,float(soc)))
    return full*(.65*(.9-s)/.9+.35) if s<.9 else full*.35*(1-s)/.1

def relay_energy(st,service_s):
    f=ST[st]
    return f['ego']+f['eback']+(R['P_hover']+R['P_comm'])*(R['t_link']+service_s)/3600

def build_sessions(demand_frame):
    # 每个站点独立；短间隙保持在站，超过返仓+周转+准备+再去程时重开会话。
    sessions=[]
    for st,d in demand_frame.groupby('中继站'):
        f=ST[st]; d=d.sort_values(['开始_s','结束_s'])
        gap_max=f['tback']+R['t_turn']+R['t_prep']+f['tgo']+R['t_link']
        blocks=[]
        for v in d.itertuples(index=False):
            a=float(v.开始_s);b=float(v.结束_s)
            if not blocks or a>blocks[-1][1]+gap_max+1e-9 or max(blocks[-1][1],b)-blocks[-1][0]>f['maxservice']+1e-9:
                blocks.append([a,b,[v.架次]])
            else:
                blocks[-1][1]=max(blocks[-1][1],b);blocks[-1][2].append(v.架次)
        for a,b,pids in blocks:
            if b-a>f['maxservice']+1e-7: raise ValueError(f'{st} 单窗超出续航')
            dep=a-R['t_prep']-f['tgo']-R['t_link']
            if dep < -1e-7: raise ValueError(f'{st} 会话无法在首个盲区窗前到达')
            dep=max(0.,dep);ret=b+f['tback'];energy=relay_energy(st,b-a)
            soc=1-energy/R['E_use']
            if energy>R['budget']+1e-7: raise ValueError(f'{st} 会话超出返航能量预算')
            sessions.append(dict(station=st,t_dep=dep,t_linked=a,t_leave=b,t_end=ret,E=energy,soc=soc,pids=tuple(sorted(set(pids)))))
    return sorted(sessions,key=lambda x:(x['t_dep'],x['station']))

from functools import lru_cache
@lru_cache(maxsize=None)
def group_eval(area_tuple):
    aset=set(area_tuple)
    sor=plan[plan.areas.apply(lambda ar:set(ar).issubset(aset))].copy()
    bx=box_df[box_df['服务区'].isin(aset)]
    dv=dem_iv[dem_iv['架次'].isin(sor['架次'])]
    sess=build_sessions(dv) if len(dv) else []
    res={'areas':tuple(sorted(aset)),'sorties':sor['架次'].tolist(),'n_sorties':len(sor),'n_boxes':len(bx),
         'mass_kg':float(bx['质量_kg'].sum()),'pop':int(areas.loc[list(aset),'pop'].sum()),
         'work_s':float(sor['持续_s'].sum()),'E_T':float(sor['能耗_kWh'].sum()),
         'E_R':float(sum(v['E'] for v in sess)),'sessions':sess}
    for g in TYPES:
        d=sor[sor['机型']==g]
        res[('UAV',g)],res[('UAV',g,'peak_t')]=max_overlap(zip(d['开始_s'],d['返仓_s']))
        res[('BAT',g)],res[('BAT',g,'peak_t')]=max_overlap(zip(d['开始_s'],d['电池再可用_s']))
        res[('UAV',g,'lanes')]=lanes([(v.开始_s,v.返仓_s,v.架次) for v in d.itertuples(index=False)])
        res[('BAT',g,'lanes')]=lanes([(v.开始_s,v.电池再可用_s,v.架次) for v in d.itertuples(index=False)])
    res['RELAY'],res[('RELAY','peak_t')]=max_overlap((v['t_dep'],v['t_end']+R['t_turn']) for v in sess)
    res['COMP'],res[('COMP','peak_t')]=max_overlap((v['t_dep'],v['t_end']+t_chg(v['soc'],T_FULL_R)) for v in sess)
    res['relay_lanes']=lanes([(v['t_dep'],v['t_end']+R['t_turn'],i) for i,v in enumerate(sess)])
    res['makespan']=max([0.]+sor['返仓_s'].tolist()+[v['t_end'] for v in sess])
    return res
JOINT=group_eval(tuple(AREA_IDS))
q3_actual={**{('UAV',g):int(plan[plan['机型']==g]['无人机'].nunique()) for g in TYPES},
           **{('BAT',g):int(plan[plan['机型']==g]['电池'].nunique()) for g in TYPES},
           'RELAY':int(q3_sess['中继无人机'].nunique()), 'COMP':int(q3_sess['组件'].nunique())}
print('全部架次联合执行的最小并发配置：','，'.join(f'{RES_LABEL[k]} {JOINT[k]}' for k in RES_KEYS))
print('Q3 原方案实际配置：','，'.join(f'{RES_LABEL[k]} {q3_actual[k]}' for k in RES_KEYS))
display(pd.DataFrame([{'资源':RES_LABEL[k],'联合峰值下界':JOINT[k],'Q3实际投用':q3_actual[k],'现有库存':INV[k]} for k in RES_KEYS]))
```

## 9 分区枚举、指标计算与推荐分区

在原子单元层面精确枚举全部 2 组与 3 组划分，逐一核算资源需求。按“加权缺口 → 冗余 → 均衡偏差”的字典序选出推荐方案；图 2 展示各划分在配置规模—均衡平面上的分布，颜色表示加权缺口，点面积表示冗余。

```python
def set_partitions(items,K):
    n=len(items)
    def rec(i,groups):
        if i==n:
            if len(groups)==K:yield tuple(tuple(g) for g in groups)
            return
        if len(groups)+(n-i)<K:return
        for g in groups:
            g.append(items[i]);yield from rec(i+1,groups);g.pop()
        if len(groups)<K:
            groups.append([items[i]]);yield from rec(i+1,groups);groups.pop()
    yield from rec(0,[])
W_GAP={**{('UAV',g):3 for g in TYPES},**{('BAT',g):1 for g in TYPES},'RELAY':3,'COMP':1}
def partition_eval(parts):
    groups=[group_eval(tuple(sorted(a for u in part for a in UNITS[u]))) for part in parts]
    total={k:sum(r[k] for r in groups) for k in RES_KEYS}
    gap={k:max(0,total[k]-INV[k]) for k in RES_KEYS}
    red={k:total[k]-JOINT[k] for k in RES_KEYS}
    works=np.array([r['work_s'] for r in groups]);shares=works/works.sum()
    beta=float(np.max(np.abs(shares-1/len(groups))))
    w_gap=sum(W_GAP[k]*gap[k] for k in RES_KEYS)
    scale=sum(total.values());red_total=sum(red.values())
    return dict(groups=groups,parts=parts,total=total,gap=gap,red=red,shares=shares,beta=beta,
                w_gap=w_gap,n_gap=sum(gap.values()),scale=scale,red_total=red_total,
                key=(w_gap,red_total,beta,scale))
PART={}
for K in (2,3):
    rows=[partition_eval(p) for p in set_partitions(tuple(range(len(UNITS))),K)]
    rows.sort(key=lambda e:(e['key'],e['parts']));PART[K]=rows
    print(f'K={K}：精确枚举 {len(rows)} 种；推荐 {rows[0]["parts"]}，加权缺口 {rows[0]["w_gap"]}，冗余 {rows[0]["red_total"]}，β={rows[0]["beta"]:.3f}')
assert len(PART[2])==2**(len(UNITS)-1)-1
BEST={K:PART[K][0] for K in (2,3)}
BAL={K:min(PART[K],key=lambda e:(e['beta'],e['w_gap'],e['red_total'],e['scale'])) for K in (2,3)}
def pname(e):return ' | '.join('{'+','.join(f'U{i+1}' for i in part)+'}' for part in e['parts'])
def rank_table(rows):
    return pd.DataFrame([{'分区':pname(e),'配置规模':e['scale'],'冗余':e['red_total'],'加权缺口':e['w_gap'],
                          '缺口件数':e['n_gap'],'均衡偏差β':round(e['beta'],3),
                          '工作量份额':'/'.join(f'{s:.0%}' for s in e['shares']),
                          **{RES_LABEL[k]:e['total'][k] for k in RES_KEYS}} for e in rows])
for K in (2,3):
    print(f'K={K} 全部划分（按推荐字典序）：');display(rank_table(PART[K]))
    print('均衡优先方案：',pname(BAL[K]))

# 图 2：所有可行分区在配置规模—均衡平面的分布
fig,axs=plt.subplots(1,2,figsize=(13,5))
for ax,K in zip(axs,(2,3)):
    rows=PART[K];xx=[e['scale'] for e in rows];yy=[e['beta'] for e in rows]
    col=[e['w_gap'] for e in rows];size=[70+22*e['red_total'] for e in rows]
    sc=ax.scatter(xx,yy,c=col,s=size,cmap='coolwarm',edgecolors='white',alpha=.8)
    ax.scatter([BEST[K]['scale']],[BEST[K]['beta']],marker='*',s=360,c=PAL['gold'],edgecolors=PAL['ink'],zorder=5,label='推荐')
    ax.scatter([BAL[K]['scale']],[BAL[K]['beta']],marker='P',s=150,c=PAL['teal'],edgecolors='white',zorder=5,label='均衡优先')
    ax.set_title(f'( {"ab"[K-2]} ) K={K}：{len(rows)} 种');ax.set_xlabel('配置规模（件/架/组）');ax.set_ylabel('工作量均衡偏差 β')
    ax.legend(loc='best');fig.colorbar(sc,ax=ax,label='加权缺口')
fig.suptitle('图 2  全部划分的配置规模、均衡与缺口');fig.tight_layout();savefig(fig,'图02_分区枚举全景')
```

## 10 推荐分区的资源配置方案与图 3、图 4

分别列出推荐方案 A（缺口优先）与均衡方案 B。图 3 为四种候选方案的分区地图，展示组色区域、组内航线与所需中继站；图 4 按资源类型堆叠各组需求，以竖线标出库存，并用红色边框标注超出部分。

```python
def config_table(K,e):
    rows=[]
    for i,r in enumerate(e['groups']):
        rows.append(dict(任务组=f'G{i+1}',原子单元=','.join(f'U{u+1}' for u in e['parts'][i]),
                         服务区='、'.join(r['areas']),服务区数=len(r['areas']),架次=r['n_sorties'],货箱=r['n_boxes'],
                         人口=r['pop'],质量_kg=r['mass_kg'],运输能耗_kWh=r['E_T'],中继能耗_kWh=r['E_R'],
                         中继会话=len(r['sessions']),完成_h=r['makespan']/3600,
                         **{RES_LABEL[k]:r[k] for k in RES_KEYS}))
    rows.append(dict(任务组='合计',服务区数=15,架次=sum(r['n_sorties'] for r in e['groups']),货箱=80,
                     人口=sum(r['pop'] for r in e['groups']),质量_kg=sum(r['mass_kg'] for r in e['groups']),
                     运输能耗_kWh=sum(r['E_T'] for r in e['groups']),中继能耗_kWh=sum(r['E_R'] for r in e['groups']),
                     中继会话=sum(len(r['sessions']) for r in e['groups']),
                     **{RES_LABEL[k]:e['total'][k] for k in RES_KEYS}))
    for name,vals in [('库存',INV),('缺口',e['gap']),('联合执行峰值',JOINT),('分区冗余',e['red'])]:
        rows.append(dict(任务组=name,**{RES_LABEL[k]:vals[k] for k in RES_KEYS}))
    return pd.DataFrame(rows).set_index('任务组')
CFG={}
for K in (2,3):
    for tag,e in [('A',BEST[K]),('B',BAL[K])]:
        CFG[K,tag]=config_table(K,e)
        print(f'K={K} 方案{tag}：{pname(e)}');display(CFG[K,tag].round(2))

# 图 3：四种候选方案的分区地图
fig,axes=plt.subplots(2,2,figsize=(15,12),constrained_layout=True)
for ax,(K,tag) in zip(axes.ravel(),[(2,'A'),(2,'B'),(3,'A'),(3,'B')]):
    e=BEST[K] if tag=='A' else BAL[K];base_map(ax)
    for gi,r in enumerate(e['groups']):
        col=GROUP_COLORS[gi];hull_patch(ax,r['areas'],col)
        for pid in r['sorties']:
            ar=plan.set_index('架次').loc[pid,'areas']
            nodes=[(LON_O,LAT_O)]+[(float(areas.loc[a,'lon']),float(areas.loc[a,'lat'])) for a in ar]+[(LON_O,LAT_O)]
            ax.plot([p[0] for p in nodes],[p[1] for p in nodes],color=col,alpha=.22,lw=.8,zorder=2)
        for a in r['areas']:
            xy=areas.loc[a];ax.scatter(xy.lon,xy.lat,s=75,color=col,edgecolors='white',zorder=5)
            ax.annotate(a,(xy.lon,xy.lat),xytext=(4,3),textcoords='offset points',fontsize=6.5)
    for v in station.itertuples(index=False):
        ax.scatter(v.经度,v.纬度,marker='D',s=52,facecolor='white',edgecolor=PAL['ink'],zorder=7)
        ax.annotate(v.中继站,(v.经度,v.纬度),xytext=(4,4),textcoords='offset points',fontsize=6.5)
    ax.set_title(f'K={K} 方案{tag}：{pname(e)}',fontsize=9)
fig.suptitle('图 3  推荐分区与均衡分区：既定航线、服务区和中继站',fontsize=12)
savefig(fig,'图03_推荐分区地图')
```

```python
# 图 4：各组需求堆叠、库存线与分区缺口
fig,axes=plt.subplots(2,2,figsize=(15,10),constrained_layout=True)
for ax,(K,tag) in zip(axes.ravel(),[(2,'A'),(2,'B'),(3,'A'),(3,'B')]):
    e=BEST[K] if tag=='A' else BAL[K];y=np.arange(len(RES_KEYS));left=np.zeros(len(y))
    for gi,r in enumerate(e['groups']):
        v=np.array([r[k] for k in RES_KEYS]);ax.barh(y,v,left=left,color=GROUP_COLORS[gi],label=f'G{gi+1}',height=.62)
        for j,z in enumerate(v):
            if z:ax.text(left[j]+z/2,j,str(z),ha='center',va='center',color='white',weight='bold',fontsize=8)
        left+=v
    for j,k in enumerate(RES_KEYS):
        ax.plot([INV[k],INV[k]],[j-.4,j+.4],color=PAL['ink'],lw=2.3)
        if e['gap'][k]:ax.text(left[j]+.15,j,f'缺 {e["gap"][k]}',va='center',color=PAL['coral'])
    ax.set_yticks(y,[RES_LABEL[k] for k in RES_KEYS]);ax.invert_yaxis();ax.set_xlim(0,max(left.max(),max(INV.values()))+2)
    ax.set_title(f'K={K} 方案{tag}：加权缺口 {e["w_gap"]}，冗余 {e["red_total"]}，β={e["beta"]:.3f}',fontsize=9)
    ax.legend(loc='lower right',ncol=3);ax.set_xlabel('设备 / 电池 / 组件数量')
fig.suptitle('图 4  各组独立执行的资源天平（黑线：库存）',fontsize=12)
savefig(fig,'图04_资源天平')
```

## 11 图 5、图 6：分组资源占用时序与独立执行甘特图

图 5 对 $K=3$ 推荐分区逐组绘制运输机在用数、电池占用数（任务加充电），峰值线对应该组配置需求。不同组的峰值即使发生在不同时间，也不能跨组调配。图 6 将运输架次排入最少资源泳道，并显示中继会话的建链、服务与返程时段。

```python
def occupancy(intervals,t):
    ans=np.zeros(len(t),int)
    for a,b in intervals:ans+=(t>=float(a))&(t<float(b))
    return ans
K_SHOW=3;e=BEST[K_SHOW]
t_end=max(r['makespan'] for r in e['groups'])+300
tg=np.arange(0,t_end+20,20.)
fig,axes=plt.subplots(K_SHOW,2,figsize=(15,9),sharex=True,constrained_layout=True)
for gi,r in enumerate(e['groups']):
    d=plan[plan['架次'].isin(r['sorties'])]
    for col,typ in enumerate(['UAV','BAT']):
        ax=axes[gi,col];ys=[]
        for g in TYPES:
            q=d[d['机型']==g]
            stop='返仓_s' if typ=='UAV' else '电池再可用_s'
            ys.append(occupancy(zip(q['开始_s'],q[stop]),tg))
        ax.stackplot(tg/3600,ys,step='post',colors=[TYPE_COLORS[g] for g in TYPES],alpha=.75,labels=[f'{g}型' for g in TYPES])
        peak=sum(r[(typ,g)] for g in TYPES)
        ax.set_ylim(0,peak+1.5);ax.set_ylabel(f'G{gi+1} 占用数')
        ax.text(.01,.95,'配置 '+ '、'.join(f'{g}:{r[(typ,g)]}' for g in TYPES),transform=ax.transAxes,va='top',fontsize=8)
        if gi==0:
            ax.set_title('(a) 在用运输无人机' if typ=='UAV' else '(b) 占用共享电池（任务 + 充满）')
            ax.legend(loc='upper right',ncol=3)
        if typ=='BAT':
            ax.axhline(sum(INV[('BAT',g)] for g in TYPES),ls='--',lw=.8,color=PAL['ink'],alpha=.35)
for ax in axes[-1]:ax.set_xlabel('灾后起算时间 / h')
fig.suptitle('图 5  K=3 推荐分区各组资源占用时序')
savefig(fig,'图05_分组资源占用时序')
```

## 13 附加分析：允许组内平移架次开始时刻，缺口能消解多少（图 8）

正式分区结果保持问题三的固定时序。附加分析保持每个架次的路线、组批、持续时间和逐箱相对送达偏移不变，只在组内按硬时限最迟开始时刻排序，并使用可用运输机与满电电池列表调度。每组完成时间最多比固定时序晚 30 min；架次平移后同步平移其通信需求，再按相同规则重建中继会话。对资源数量逐类下降，得到经过核验的可行配置上界；这不证明所有平移方案中的严格最小值。图 8 比较固定时序需求、平移后可行需求与库存。

```python
# 排序与释放：通信接入必须在进入盲区前完成。
plan_by_id=plan.set_index('架次')
box_by_pid={pid:d for pid,d in box_df.groupby('架次')}
demand_by_pid={pid:d for pid,d in dem_iv.groupby('架次')}
def latest_start(pid):
    d=box_by_pid[pid].dropna(subset=['硬时限_s'])
    if d.empty:return float('inf')
    original=float(plan_by_id.loc[pid,'开始_s'])
    return float((d['硬时限_s']-(d['送达_s']-original)).min())
def earliest_start(pid):
    if pid not in demand_by_pid:return 0.
    d=demand_by_pid[pid];original=float(plan_by_id.loc[pid,'开始_s'])
    return max(0.,max(R['t_prep']+ST[st]['tgo']+R['t_link']-(a-original)
                      for a,st in zip(d['开始_s'],d['中继站'])))
LST={p:latest_start(p) for p in plan['架次']}
EST={p:earliest_start(p) for p in plan['架次']}

def reschedule_group(r,cfg):
    sub=plan[plan['架次'].isin(r['sorties'])]
    # 两种确定性顺序；资源列表调度可提前到 t=0，绝不推迟硬时限。
    orders=[sub.sort_values(['开始_s','架次'])['架次'].tolist(),
            sorted(sub['架次'],key=lambda p:(LST[p],float(plan_by_id.loc[p,'开始_s']),p))]
    best=None
    for order in orders:
        u={g:[0.]*cfg[('UAV',g)] for g in TYPES};b={g:[0.]*cfg[('BAT',g)] for g in TYPES}
        starts={};rows=[];valid=True
        for pid in order:
            v=plan_by_id.loc[pid];g=v['机型']
            if not u[g] or not b[g]:valid=False;break
            i=min(range(len(u[g])),key=lambda i:u[g][i]);j=min(range(len(b[g])),key=lambda j:b[g][j])
            st=max(u[g][i],b[g][j],EST[pid]);en=st+float(v['持续_s'])
            if st>LST[pid]+1e-7:valid=False;break
            u[g][i]=en;b[g][j]=en+float(v['充电_s'])
            starts[pid]=st;rows.append((pid,g,i,j,st,en,b[g][j]))
        if not valid:continue
        dv=dem_iv[dem_iv['架次'].isin(starts)].copy()
        dv['shift_s']=dv['架次'].map(starts)-dv['架次'].map(plan_by_id['开始_s'])
        dv['开始_s']+=dv['shift_s'];dv['结束_s']+=dv['shift_s']
        try:sessions=build_sessions(dv) if len(dv) else []
        except ValueError:continue
        nr=max_overlap((x['t_dep'],x['t_end']+R['t_turn']) for x in sessions)[0]
        nc=max_overlap((x['t_dep'],x['t_end']+t_chg(x['soc'],T_FULL_R)) for x in sessions)[0]
        mk=max([v[5] for v in rows]+[x['t_end'] for x in sessions]+[0.])
        if mk>r['makespan']+1800+1e-7 or nr>r['RELAY'] or nc>r['COMP']:continue
        cand=dict(starts=starts,rows=rows,sessions=sessions,makespan=mk,relay=nr,comp=nc)
        if best is None or (mk,nr,nc)<(best['makespan'],best['relay'],best['comp']):best=cand
    return best

def reduce_config(r):
    cfg={k:r[k] for k in RES_KEYS}
    current=reschedule_group(r,cfg)
    if current is None:raise RuntimeError('固定配置在平移列表调度下不可行')
    changed=True
    while changed:
        changed=False
        for key in [('UAV',g) for g in TYPES]+[('BAT',g) for g in TYPES]:
            if cfg[key]==0:continue
            trial={**cfg,key:cfg[key]-1}
            cand=reschedule_group(r,trial)
            if cand is not None:cfg=trial;current=cand;changed=True
    # 后续重新建链得到的中继/组件峰值，是此可行时序的实际需求。
    cfg['RELAY']=current['relay'];cfg['COMP']=current['comp']
    return cfg,current
SHIFT={}
for K in (2,3):
    rows=[]
    for gi,r in enumerate(BEST[K]['groups']):
        cfg,sol=reduce_config(r);SHIFT[K,gi]=(cfg,sol)
        rows.append({'任务组':f'G{gi+1}','固定完成_h':r['makespan']/3600,'平移完成_h':sol['makespan']/3600,
                     **{f'{RES_LABEL[k]} 固定':r[k] for k in RES_KEYS},
                     **{f'{RES_LABEL[k]} 平移':cfg[k] for k in RES_KEYS}})
    print(f'K={K} 各组平移后可行配置：');display(pd.DataFrame(rows).set_index('任务组').round(2))
    old=BEST[K]['total'];new={k:sum(SHIFT[K,i][0][k] for i in range(K)) for k in RES_KEYS}
    print('库存缺口：固定时序', {RES_LABEL[k]:old[k]-INV[k] for k in RES_KEYS if old[k]>INV[k]},
          '→ 平移后可行上界', {RES_LABEL[k]:new[k]-INV[k] for k in RES_KEYS if new[k]>INV[k]})

# 图 8：固定时序资源需求、组内平移后的可行需求与库存。
fig,axes=plt.subplots(1,2,figsize=(15,6),sharey=True)
for ax,K in zip(axes,(2,3)):
    y=np.arange(len(RES_KEYS));fixed=np.array([BEST[K]['total'][k] for k in RES_KEYS])
    shifted=np.array([sum(SHIFT[K,i][0][k] for i in range(K)) for k in RES_KEYS]);inv=np.array([INV[k] for k in RES_KEYS])
    for yi in y:ax.plot([shifted[yi],fixed[yi]],[yi,yi],color=PAL['grey'],lw=3,alpha=.55,zorder=1)
    ax.scatter(fixed,y,color=PAL['coral'],s=85,label='固定时序',zorder=3)
    ax.scatter(shifted,y,color=PAL['teal'],s=85,label='组内平移可行值',zorder=4)
    ax.scatter(inv,y,color=PAL['ink'],marker='|',s=280,label='库存',zorder=5)
    ax.set_yticks(y,[RES_LABEL[k] for k in RES_KEYS]);ax.invert_yaxis()
    ax.set_xlim(-.2,max(fixed.max(),shifted.max(),inv.max())+2);ax.set_xlabel('件 / 架 / 组')
    ax.set_title(f'K={K}：各组完成时间 '+', '.join(f'G{i+1} {SHIFT[K,i][1]["makespan"]/3600:.2f}h' for i in range(K)),fontsize=9)
axes[0].legend(loc='lower right',fontsize=8)
fig.suptitle('图 8  组内时刻平移对推荐分区资源缺口的影响（30 min 完成时限放宽）')
fig.tight_layout();savefig(fig,'图08_平移调度消解缺口')
```

## 14 结果导出与结论

最终核对服务区覆盖、架次归组、多点架次不可拆分、资源泳道峰值、中继需求覆盖与能量返航约束，以及平移后的硬时限和 30 min 完成时间限制。下方导出全部划分、推荐配置、比较、缺口归因和核验结果，并汇总两种分区的主要缺口。结论仅适用于本 Notebook 读取的问题三方案快照及固定时序核算口径。

```python
checks=[]
def check(name,ok,detail=''):checks.append(dict(检查项=name,通过=bool(ok),说明=detail))
for K in (2,3):
    v=BEST[K]
    all_areas=[a for r in v['groups'] for a in r['areas']]
    all_pid=[p for r in v['groups'] for p in r['sorties']]
    check(f'K={K} 每区仅一组',Counter(all_areas)==Counter(AREA_IDS))
    check(f'K={K} 每架次仅一组',Counter(all_pid)==Counter(plan['架次']))
    check(f'K={K} 多点架次不拆开',all(set(ar).issubset(set(next(r['areas'] for r in v['groups'] if a in r['areas'])))
                                      for ar in plan.areas for a in ar[:1]))
    check(f'K={K} 每组资源泳道数等于峰值',all(len(set(r[('UAV',g,'lanes')].values()))==r[('UAV',g)] and
                                             len(set(r[('BAT',g,'lanes')].values()))==r[('BAT',g)] for r in v['groups'] for g in TYPES))
    covered=True;energy=True
    for r in v['groups']:
        d=dem_iv[dem_iv['架次'].isin(r['sorties'])]
        for w in d.itertuples(index=False):
            covered &= any(x['station']==w.中继站 and x['t_linked']<=w.开始_s+1e-7 and
                           x['t_leave']>=w.结束_s-1e-7 for x in r['sessions'])
        energy &= all(x['E']<=R['budget']+1e-7 and x['soc']>=R['rho']-1e-7 for x in r['sessions'])
    check(f'K={K} 同站中继窗全覆盖',covered,f'{len(dem_iv)} 个原始需求窗按组分配')
    check(f'K={K} 中继能量与返航余量',energy)
    for gi,r in enumerate(v['groups']):
        cfg,sol=SHIFT[K,gi]
        offset=box_df[box_df['架次'].isin(sol['starts'])].copy()
        offset['new_arrival']=offset['送达_s']+offset['架次'].map(sol['starts'])-offset['架次'].map(plan_by_id['开始_s'])
        hard=offset.dropna(subset=['硬时限_s'])
        check(f'K={K} G{gi+1} 平移硬时限',bool((hard['new_arrival']<=hard['硬时限_s']+1e-7).all()),f'{len(hard)} 箱')
        check(f'K={K} G{gi+1} 完成时间',sol['makespan']<=r['makespan']+1800+1e-7)
        check(f'K={K} G{gi+1} 平移中继资源',sol['relay']<=r['RELAY'] and sol['comp']<=r['COMP'])
VER=pd.DataFrame(checks);display(VER)
print('核对结果：', '全部通过' if VER['通过'].all() else '存在未通过项，请查看上表')

# 截图中的结果表对应本机 Q3 方案，便于论文附录引用；原始附件不改写。
xlsx_path=CODE/'问题四/问题四结果.xlsx'
with pd.ExcelWriter(xlsx_path,engine='openpyxl') as writer:
    unit_df.to_excel(writer,sheet_name='1_原子单元',index=False)
    rank_table(PART[2]).to_excel(writer,sheet_name='2_K2全部划分',index=False)
    rank_table(PART[3]).to_excel(writer,sheet_name='3_K3全部划分',index=False)
    for K in (2,3):
        CFG[K,'A'].to_excel(writer,sheet_name=f'4_K{K}推荐配置')
        CFG[K,'B'].to_excel(writer,sheet_name=f'5_K{K}均衡配置')
    cmp_df.to_excel(writer,sheet_name='6_分区对比')
    cause_df.to_excel(writer,sheet_name='7_缺口归因',index=False)
    VER.to_excel(writer,sheet_name='8_可行性核对',index=False)
print('结果：',xlsx_path.relative_to(PROJECT))
for K in (2,3):
    v=BEST[K]
    print(f'K={K} 推荐 {pname(v)}；配置 {v["scale"]}，冗余 {v["red_total"]}，β={v["beta"]:.3f}，加权缺口 {v["w_gap"]}')
    print('  资源缺口：', {RES_LABEL[k]:v['gap'][k] for k in RES_KEYS if v['gap'][k]>0})
    shifted={k:sum(SHIFT[K,i][0][k] for i in range(K)) for k in RES_KEYS}
print('  平移后可行配置缺口：',{RES_LABEL[k]:max(0,shifted[k]-INV[k]) for k in RES_KEYS if shifted[k]>INV[k]})
```
